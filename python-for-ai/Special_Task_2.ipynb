```python
df['is_canceled'].value_counts()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data = df , x = "is_canceled" , palette = "Set2")
plt.title("Booking Cancellation Status")
plt.xlabel("Booking Status")
plt.ylabel("No.of.Bookings")
plt.xticks([0,1],["Not Cancelled","Cancelled"])
plt.show()
```

```python
df["hotel"].value_counts()
```

```python
plt.figure(figsize=(8,6))

sns.countplot(data = df , x = "hotel" , palette = "pastel")
plt.title("Booking by Hotel Type")
plt.xlabel("Hotel Types")
plt.ylabel("No.of.Bookings")

plt.show()
```

```python
hotel_counts = df["hotel"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(hotel_counts,labels=hotel_counts.index,autopct="%1.1f%%",startangle=90,explode=(0.05,0),shadow=True)

plt.title("Percentage of Bookings by Hotel Type")

plt.show()
```

```python
df["arrival_date_month"].value_counts()
```

```python
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December"]

plt.figure(figsize=(12,5))

sns.countplot( data=df, x="arrival_date_month",order = month_order,palette="viridis")

plt.title("Bookings by Month")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=45)

plt.show()
```

```python
df['lead_time'].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,x="lead_time",bins=30,kde=True,color="skyblue")

plt.title("Lead Time Distribution")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Frequency")

plt.show()
```

```python
df['adr'] = df['adr'].fillna(0)
```

```python
df["adr"].max(0)
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(data=df,y="adr",color="lightgreen")

plt.title("Average Daily Rate Distribution")
plt.ylim(-10, 600)
plt.show()
```

```python
df.describe()
```

```python

df[df["adr"] == 5400]
```

```python

df = df[df["adr"] < 5000]
```

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot( data=df, x="is_canceled", y="lead_time", palette="Set2")

plt.title("Lead Time vs Booking Cancellation")

plt.xticks([0,1],["Not Cancelled","Cancelled"])

plt.show()
```

```python
top_countries = df["country"].value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))

ax = sns.barplot( x=top_countries.index, y=top_countries.values,palette="mako")

plt.title("Top 10 Countries by Number of Bookings", fontsize=15)
plt.xlabel("Country")
plt.ylabel("Number of Bookings")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df,x="lead_time",y="adr", alpha=0.5)

plt.title("Lead Time vs Average Daily Rate")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Average Daily Rate")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="deposit_type",
    hue="is_canceled",
    palette="Set2"
)

plt.title("Booking Cancellation by Deposit Type")
plt.xlabel("Deposit Type")
plt.ylabel("Number of Bookings")

plt.show()
```
