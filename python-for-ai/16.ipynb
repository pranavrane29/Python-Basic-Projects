# ***Identifying Numerical & Categorical Data:***

```python
df.columns
```

```python
print(f"The column age has numerical data - ({df["age"].dtype})")
print(f"The column sex has categorical data - ({df["sex"].dtype})")
print(f"The column bmi has numerical data - ({df["bmi"].dtype})")
print(f"The column children has numerical data - ({df["children"].dtype})")
print(f"The column smoker has categorical data - ({df["smoker"].dtype})")
print(f"The column region has categorical data - ({df["region"].dtype})")
print(f"The column expenses has numerical data - ({df["expenses"].dtype})")
```

# ***Distribution plots for all numerical values:***

```python
#Plot for Age Column -
plt.figure(figsize=(8,6))

sns.histplot(data=df,x="age",color="skyblue",kde=True)
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

```python
#Plot for BMI Column -
plt.figure(figsize=(8,6))

sns.histplot(data=df,x="bmi",color="pink",kde=True)
plt.title("BMI distribution")
plt.xlabel("BMI")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

```python
#Plot for Children Column -
plt.figure(figsize=(8,6))

sns.histplot(data=df,x="children",color="lightgreen",kde=True)
plt.title("Child distribution")
plt.xlabel("Age")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

```python
#Plot for Expenses Column -
plt.figure(figsize=(8,6))

sns.histplot(data=df,x="expenses",color="cyan",kde=True)
plt.title("Expense distribution")
plt.xlabel("Expenses")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

# ***Count Plots for all the categorical columns:***

```python
#Plot for Sex Column -
plt.figure(figsize=(8,6))

sns.countplot(data=df,x="sex",palette="Set2",edgecolor = "Black")
plt.title("Gender Distribution")
plt.xlabel("Sex")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

```python
#Plot for Sex Column -
plt.figure(figsize=(8,6))

sns.countplot(data=df,x="smoker",palette="Set3",edgecolor = "Black")
plt.title("Smoker's Distribution")
plt.xlabel("Is a Smoker?")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

```python
#Plot for Region Column -
plt.figure(figsize=(8,6))

sns.countplot(data=df,x="region",palette="Set1",edgecolor = "Black")
plt.title("Regional Distributions")
plt.xlabel("Regions")
plt.ylabel("No.of.counts")
plt.grid()
plt.show()
```

# ***Correlation between numerical variables(heatmap):***

```python
corr_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))

sns.heatmap(corr_matrix,annot=True,cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
