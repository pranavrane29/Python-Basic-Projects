```python
from dotenv import load_dotenv
import os
from langchain_groq import ChatGroq

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")

model = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=groq_api_key
)


def structured_response(topic):
    prompt = f"""
You are a helpful AI tutor.

Explain the following topic: {topic}

Provide the answer in exactly this structure:

1. Definition:
Give a short and simple definition.

2. Three Key Points:
- Point 1
- Point 2
- Point 3

3. Real-Life Example:
Give one simple real-life example.

Keep the explanation clear and easy to understand.
"""

    response = model.invoke(prompt)
    return response.content


result = structured_response("Machine Learning")

print(result)
```

```python
from dotenv import load_dotenv
import os
from langchain_groq import ChatGroq
from langchain_mistralai import ChatMistralAI

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
mistral_api_key = os.getenv("MISTRAL_API_KEY")

if not groq_api_key and not mistral_api_key:
    print("Error: No API keys found. Please check your .env file.")
else:

    groq_model = None
    mistral_model = None

    if groq_api_key:
        try:
            groq_model = ChatGroq(
                model="llama-3.1-8b-instant",
                api_key=groq_api_key
            )
        except Exception as e:
            print("Error initializing Groq model:", e)

    if mistral_api_key:
        try:
            mistral_model = ChatMistralAI(
                model="mistral-small-2603",
                api_key=mistral_api_key
            )
        except Exception as e:
            print("Error initializing Mistral model:", e)

    print("===== Multi-Model AI Assistant =====")
    print("Choose Groq or Mistral.")
    print("Type 'quit' to exit.\n")

    while True:

        choice = input("Choose model (Groq/Mistral): ").strip().lower()

        if choice == "quit":
            print("Assistant: Goodbye!")
            break

        if choice not in ["groq", "mistral"]:
            print("Invalid choice. Please choose Groq or Mistral.\n")
            continue

        
        question = input("Enter your question: ").strip()

        if question.lower() == "quit":
            print("Assistant: Goodbye!")
            break

        if not question:
            print("Please enter a question.\n")
            continue

        if choice == "groq":
            selected_model = groq_model
            model_name = "Groq"

        else:
            selected_model = mistral_model
            model_name = "Mistral"

        if selected_model is None:
            print(f"Error: {model_name} model is not available.\n")
            continue

        try:
            response = selected_model.invoke(question)

            print(f"\n{model_name} Response:")
            print(response.content)
            print()

        except Exception as e:
            print(f"Error while getting response from {model_name}:")
            print(e)
            print()
```
